# Data Exploration: BTCUSDT Perpetual

Explore raw market data: klines, funding rates, open interest, and basic statistics.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('darkgrid')
plt.rcParams['figure.figsize'] = (14, 6)
```

## Load Canonical Dataset

```python
from crypto_quant.data.loaders import load_canonical_dataset

data = load_canonical_dataset()
print(f"Dataset shape: {data.shape}")
print(f"Date range: {data.index[0]} to {data.index[-1]}")
data.head()
```

## Summary Statistics

```python
data.describe()
```

## Price Chart

```python
plt.figure(figsize=(14, 6))
plt.plot(data.index, data['close'], linewidth=0.5)
plt.title('BTCUSDT Close Price')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Volume Analysis

```python
# Resample to daily for cleaner visualization
daily_volume = data['volume'].resample('D').sum()

plt.figure(figsize=(14, 4))
plt.bar(daily_volume.index, daily_volume.values, width=0.8)
plt.title('Daily Trading Volume')
plt.xlabel('Date')
plt.ylabel('Volume')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Funding Rate Analysis

```python
if 'funding_rate' in data.columns:
    plt.figure(figsize=(14, 4))
    plt.plot(data.index, data['funding_rate'], linewidth=0.5)
    plt.title('Funding Rate')
    plt.xlabel('Date')
    plt.ylabel('Funding Rate')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    print(f"Mean funding rate: {data['funding_rate'].mean():.6f}")
    print(f"Std dev: {data['funding_rate'].std():.6f}")
    print(f"Min: {data['funding_rate'].min():.6f}")
    print(f"Max: {data['funding_rate'].max():.6f}")
```

## Open Interest Analysis

```python
if 'open_interest' in data.columns:
    plt.figure(figsize=(14, 4))
    plt.plot(data.index, data['open_interest'], linewidth=0.5, color='orange')
    plt.title('Open Interest')
    plt.xlabel('Date')
    plt.ylabel('Open Interest')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

## Data Quality Checks

```python
print("Missing values:")
print(data.isnull().sum())
print(f"\nDuplicate timestamps: {data.index.duplicated().sum()}")
print(f"Total rows: {len(data)}")
```
